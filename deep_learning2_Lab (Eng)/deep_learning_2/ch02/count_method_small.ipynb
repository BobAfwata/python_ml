```python
import sys
sys.path.append('..')
import numpy as np
import matplotlib.pyplot as plt
from common.util import preprocess, create_co_matrix, ppmi
```

```python
text = 'You say goodbye and I say hello.'
corpus, word_to_id, id_to_word = preprocess(text)
vocab_size = len(id_to_word)
C = create_co_matrix(corpus, vocab_size, window_size=1)
W = ppmi(C)
```

```python
# SVD
U, S, V = np.linalg.svd(W)
```

```python
np.set_printoptions(precision=3)  # Display up to three decimal places
print(C[0])  # Display the first row of the co-occurrence matrix
print(W[0])  # Display the first row of the PPMI matrix
print(U[0])  # Display the first row of the matrix U from SVD
```

```python
# Plot
for word, word_id in word_to_id.items():
    plt.annotate(word, (U[word_id, 0], U[word_id, 1]))  # Annotate each point with its corresponding word
plt.scatter(U[:,0], U[:,1], alpha=0.5)  # Scatter plot of word vectors
plt.show()  # Show the plot
```
