```python
import sys
sys.path.append('..')
from common.optimizer import SGD
from common.trainer import RnnlmTrainer
from dataset import ptb
from ch05.simple_rnnlm import SimpleRnnlm
import matplotlib.pyplot as plt
plt.rc('font', family='NanumBarunGothic')
```

```python
# Hyperparameter
batch_size = 10
wordvec_size = 100
hidden_size = 100  # Number of elements in the hidden state vector of the RNN
time_size = 5  # Size of unfolding the RNN
lr = 0.1
max_epoch = 100
```

```python
# Read training data
corpus, word_to_id, id_to_word = ptb.load_data('train')
corpus_size = 1000  # Set the test dataset small
corpus = corpus[:corpus_size]
vocab_size = int(max(corpus) + 1)
xs = corpus[:-1]  # Input
ts = corpus[1:]   # Output (Correct labels)
```

```python
# Model Creation
model = SimpleRnnlm(vocab_size, wordvec_size, hidden_size)
optimizer = SGD(lr)
trainer = RnnlmTrainer(model, optimizer)

trainer.fit(xs, ts, max_epoch, batch_size, time_size)
trainer.plot()
```
