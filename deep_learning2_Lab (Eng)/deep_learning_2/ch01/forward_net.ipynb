```python
import numpy as np
```

```python
class Sigmoid:
    def __init__(self):
        # Initialize an empty list to store parameters
        self.params = []

    def forward(self, x):
        # Perform forward pass of sigmoid function
        return 1 / (1 + np.exp(-x))


class Affine:
    def __init__(self, W, b):
        # Initialize parameters W and b
        self.params = [W, b]

    def forward(self, x):
        # Perform forward pass of affine transformation
        W, b = self.params
        out = np.dot(x, W) + b
        return out
```

```python
class TwoLayerNet:
    def __init__(self, input_size, hidden_size, output_size):
        I, H, O = input_size, hidden_size, output_size

        # Weight and bias initialization
        W1 = np.random.randn(I, H)
        b1 = np.random.randn(H)
        W2 = np.random.randn(H, O)
        b2 = np.random.randn(O)

        # Layer creation
        self.layers = [
            Affine(W1, b1),  # Affine transformation
            Sigmoid(),       # Activation function
            Affine(W2, b2)   # Affine transformation
        ]

        # Collect all parameters into a list
        self.params = []
        for layer in self.layers:
            self.params += layer.params

    def predict(self, x):
        for layer in self.layers:
            x = layer.forward(x)
        return x
```

```python
x = np.random.randn(10, 2)
model = TwoLayerNet(2, 4, 3)
s = model.predict(x)
print(s)
```
